```python
# If we have 1000 datapoints and has 900 yes and 100 no
# it is 9:1 ratio so it is the imbalanced dataset
# if we handle imbalanced dataset then model will lean towards the highest side in machine learning
# Upsampling
# Downsampling
```

```python
#Up sampling
import numpy as np
import pandas as pd

#set the random sedd for reproducibility
np.random.seed(123)

#create the dataframe with two classes
n_samples = 1000 #Taking the class sample size is 1000
class_0_ratio = 0.9 #first class ratio is 90 %
n_class_0 = int(n_samples*class_0_ratio) #number of people in class0
n_class_1 = n_samples - n_class_0 #number of people in class 1
```

```python
n_class_0,n_class_1
```

```python
#Create my dataframe with imbalance dataset
class_0 = pd.DataFrame({
    'feature1':np.random.normal(loc=0,scale=1,size=n_class_0), #feature 1 is a column that takes random but has normal distribution with size class of 0
    'feature2':np.random.normal(loc=0,scale=1,size=n_class_0), #feature 1 is a column that takes random but has normal distribution with size class of 0
    'target':[0]*n_class_0 #target is column with all values 0 and size of class is class 0 size - 900 zeroes will be created

})
class_1 = pd.DataFrame({
    'feature1':np.random.normal(loc=0,scale=1,size=n_class_1),#feature 1 is a column that takes random but has normal distribution with size class of 1
    'feature2':np.random.normal(loc=0,scale=1,size=n_class_1), #feature 1 is a column that takes random but has normal distribution with size class of 1
    'target':[1]*n_class_1 #target is column with all values 1 and size of class is class 1 size - 100 ones will be created
})
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True) #dropping the index that is created
```

```python
df.head()
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
#upscaling
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample #resample it does create extra points to match majority
df_minority_upscaled=resample(df_minority,replace=True,n_samples=len(df_majority),random_state=42)
```
