Nominal/OHE Encoding
One hot encoding also known as nominal encoding is a technique used to represent categorical data as numerical data, which is more suitable for machine learning algorithms. In this technique, each category is represnted as a binary vector where each bit corresponds to a unique category. For example if we have a categorical variable 'color' with three possible values(red, green, blue) we can represent it one hot encoding as follows:

1. Red:[1,0,0]
2. Green:[0,1,0]
3. Blue:[0,0,1]

```python
# Disadvantages
# if we have 100 categories we will have 100 features
# sparse matrix - overfitting
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
#Create a simple dataframe
df = pd.DataFrame({'color':['red','blue','green','green','red','blue']})
```

```python
#Create an instance of Onehotencoder
encoder=OneHotEncoder()
```

```python
#Perform fit and transform
encoded = encoder.fit_transform(df[['color']]).toarray()
```

```python
import pandas as pd
encoder_df = pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
#if we add new data
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```

Label Encoding
Label encoding and ordinal encoding are two techniques used to encode categorical data as numerical data

Label encoding involves assigning a unique numerical label to each category in the variable. The labels are usually assigned in alphabetical order or based on the frequency of the categories. For example, if we have categorical variable color with three possible values(red,green,blue), we can represent it using label encoding as follows:

Red:1
Green:2
Blue:3

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
lbl_encoder.transform([['red']])
```

```python
lbl_encoder.transform([['blue']])
```

#Ordinal Encoding
It is used to encode categorical data that have intrisic order or ranking. In this technique, each category is assigned a numerical value based on its position in the order. For example if we have a categorical variable "educational level" with four possible values(high school, college, graduate, post-graduate), we can represnt it using ordinal encoding as follows:

High school:1
College: 2
Graduate: 3
Post-graduate:4

```python
#Ordinal Encoding
from sklearn.preprocessing import OrdinalEncoder
```

```python
#Ccreate a sample dataframe with an ordinal variable
df = pd.DataFrame({'size':['small','medium','large','medium','small','large']})
```

```python
df
```

```python
#Create an instance of ordinalencoder and then fit_trannsform
encoder = OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder.fit_transform(df[['size']])
```

```python
#newdata
encoder.transform([['small']])
```
