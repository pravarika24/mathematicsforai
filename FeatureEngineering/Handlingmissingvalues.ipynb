Missing values occur in dataset when some of the informations is not stored for a variable. There are 3 mechanisms

1 Missing Completely ar Random, MCAR

MCAR is a type of missing data mechanism in which the probability of a value being missing is unrelated to both the observed data and the missing data. In other words, if the data is MCAR, the missing values are randomly distributed throughout the dataset, and there is no systematic reason for why they are missing

For ex in a survey about the prevalence of a certain disease, the missing data might be MCAR if the survey participants with missing values for certain questions were selected randomly and their missing responses are not related to their disesase status or any other variables measured in the survey



2. Missing at Random MAR
MAR  is a type of missing data mechanism in which the probability of value being missing depends only on the observed data but not on the missing data itself. In other words, if the data is MAR, the missing values are systematically related to the observed data, but not to the missing data. Here are a few examples of missing at random


Income data: Suppose you are collecting income data from group of people, but some participants choose not to report their income. If the decision to report or not report income is related to participants age or gender, but not to their income level, then the data is missing at random.

Medical data: Suppose you are collecting medical data on patients, including their blood pressure, but some patients do not pressure. IF the patients who do not report their blood pressure are more likely to be younger or have healtier lifestyles. This is not related to their actual blood pressure values, then the data is mssing at random



Missing data not at random(MNAR)
It is a type of missing data mechanism where the probability of missing values depends on the value of missing data itself. In other words, if the data is MNAR, the missingness is not random and is dependent on unobserved or unmeasured factors that are associated with missing values.

For example, suppose you are collecting data on the income and job satisfaction of employees in a company. If employees in a company. If employees who are less satisfied with their jobs are more likely to refuse to report their income, then the data is not missing at random. In this case, the missigness is dependent on job satisfaction, which is not observed or measured.

```python
#Imputation missing techniques
#1- mean value imputation - replace the missing values with the mean of the cloumn
sns.histplot(df['age'])
```

```python
df['Age_mean'] = df['age'].fillna(df['age'].mean())
```

```python
df[['Age_mean','age']]
```

```python
# Mean imputation works wells when we have a normally distributed data
```

### Median value imputation. If we have outliers in the dataset

```python
df['age_median'] = df['age'].fillna(df['age'].median())
```

```python
df[['age_median','Age_mean','age']]
```

```python
#Mode imputation technique - categorical techniques
df[df['embarked'].isnull()]
```

```python
df['embarked'].unique()
```

```python
mode_value = df[df['embarked'].notna()]['embarked'].mode()[0]
```

```python
df['embarked_mode']=df['embarked'].fillna(mode_value)
```
